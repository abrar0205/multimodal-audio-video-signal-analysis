# Multimodal Audio-Video Signal Analysis Demo

This notebook demonstrates the public API using synthetic audio. It does not include or reproduce any private recordings, frames, face images, or derived results.

## Generate Synthetic Audio

```python
from pathlib import Path
from examples.generate_demo_audio import make_synthetic_voice_like_signal
from multimodal_signals.audio_io import write_wav

sample_rate = 16000
signal = make_synthetic_voice_like_signal(sample_rate=sample_rate)
path = Path('data/demo_synthetic_voice.wav')
path.parent.mkdir(exist_ok=True)
write_wav(path, signal, sample_rate)
path
```

## Inspect Signal Quality

```python
from multimodal_signals import inspect_media

inspect_media(path)
```

## Extract Acoustic Features

```python
from multimodal_signals import analyze_audio

audio_result = analyze_audio(path)
audio_result['features']
```
